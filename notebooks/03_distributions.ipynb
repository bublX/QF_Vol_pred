```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from source.utils.distribution import DistributionCalculator

calculator = DistributionCalculator()
```

```python
ticker = "SBER"
start = "2024-01-01"
end = "2024-12-31"

prices = pd.read_csv(
    ROOT / "data/raw/all_prices.csv",
    usecols=["date", f"{ticker}_close"],
    parse_dates=["date"],
    index_col="date",
).sort_index()

returns = pd.read_csv(
    ROOT / "data/processed/profit_matrix.csv",
    usecols=["date", f"{ticker}_returns"],
    parse_dates=["date"],
    index_col="date",
).sort_index()

prices = prices.loc[start:end, f"{ticker}_close"]
returns = returns.loc[start:end, f"{ticker}_returns"]

pd.concat([
    calculator.calculate(prices),
    calculator.calculate(returns),
])
```

```python
calculator.plot(
    prices,
    bins=80,
    title=f"{ticker}: распределение цен",
    xlabel="Цена, руб.",
)
plt.show()

calculator.plot(
    returns,
    bins=100,
    title=f"{ticker}: распределение доходностей",
    xlabel="Логарифмическая доходность, %",
)
plt.show()

calculator.plot_qq(
    returns,
    title=f"{ticker}: сравнение с нормальным распределением",
    ylabel="Логарифмическая доходность, %",
)
plt.show()
```
